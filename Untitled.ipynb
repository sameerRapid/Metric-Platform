```python
import pandas as pd

# Load the dataset (Replace 'your_file.csv' with the actual file path if needed)
# df = pd.read_csv('your_file.csv')

# Filtering for 'Previously_Not_Purchased_Subs'
df_filtered = df[df['Subs_Purchase_or_not'] == 'Previously_Not_Purchased_Subs']
# df_filtered = df[df['Subs_Purchase_or_not'] == 'Previously_Purchased_Subs']

# Aggregation
result = df_filtered.groupby(['city', 'Subs_Purchase_or_not']).agg(
    exp_base=('captain_id', 'nunique'),
    visited_subs_screen_caps=('captain_id_visited', 'nunique'),
    purchased_subs_caps=('subsCaps', lambda x: x[df_filtered['screenName'].isin(
        ['razorpay_screen', 'paytm_screen', 'cashfree_screen'])].nunique()),
    any_model_confirm=('user_id', 'nunique')
).reset_index()

# Calculating percentages
result['Subs_Adoption_percent'] = (result['purchased_subs_caps'] * 100.0) / result['exp_base']
result['Byob_plan_Confirmed_wrt_exp_percent'] = (result['any_model_confirm'] * 100.0) / result['exp_base']
result['Byob_plan_Confirmed_wrt_visited_percent'] = (result['any_model_confirm'] * 100.0) / result['visited_subs_screen_caps']
```

```python
def control(city,pdd,ed):
    q=f"""
       with base as (
        SELECT
        captain_id, city, performance_segment, consistency_segment
            from
             hive.experiments_internal.captain_byob_experiment_2_base_control
           where lower(city) in ('hyderabad')
        ),
        
        
        vistitedcaps as (
        select  
        updated_yyyymmdd as yyyymmdd,
        updated_epoch,
        json_extract_scalar(event_props, '$.city') AS city,
        event_props_user_id as captain_id,
        json_extract_scalar(event_props, '$.screen_name') as screenName,
        json_extract_scalar(event_props, '$.source') as ingress
        from clevertap.captain_subscription_screen_visited_immutable
        where json_extract_scalar(event_props, '$.Captain_Mode') not in ('5fbe8a8a9788ac0008c4eb98','642ae204b4b6b8ec5665ce87')
        and yyyymmdd between '{ed[0]}' and  '{ed[1]}'
        and json_extract_scalar(event_props, '$.screen_name') not in ('confirmation_sheet') -- Landing Page
        and lower(json_extract_scalar(event_props, '$.city')) in('hyderabad') 
        and event_props_user_id in (select captain_id from base)
        ),
        
        captainService as(
         select distinct captain_id
                from datasets_internal.captain_subservice_segments_v1
                where lower(service_mode)= 'all'
                and lower(service) = 'link'
                and run_date between '{ed[0]}' and '{ed[1]}'
                and captain_id in (select distinct captain_id from base)
        ),
        
        visitedService as (
         select a.yyyymmdd,a.ingress,a.city,
         a.captain_id as captain_id_visited ,a.screenName,
         row_number() over(partition by a.yyyymmdd,a.captain_id,a.city order by a.updated_epoch desc) as r
         from vistitedcaps a
         inner join captainService b
         on a.captain_id=b.captain_id
        ),
        
        subsPurchased as (
          select 
          -- distinct owner_id as captain_id
          yyyymmdd,owner_id as captain_id, epoch
          from payments.transactions_snapshot
          where lower(subtype) like '%subs%'
          and lower(owner_type)='captain'
          and yyyymmdd between '{ed[0]}' and '{ed[1]}' -- Kol
          and subtype = 'subscriptionPurchase'
        ),
        
        subs_purchased_wrt_ingress as (
        select distinct captain_id_subs, purchase_ingress
        from (
        select b.captain_id as captain_id_subs, a.ingress as purchase_ingress, a.screenName, b.epoch, 
        row_number() over(partition by b.captain_id order by epoch desc) as row 
        from visitedService a 
        left join subsPurchased b 
        on a.captain_id_visited=b.captain_id
        and a.yyyymmdd=b.yyyymmdd
        where screenName in ('razorpay_screen','paytm_screen','cashfree_screen')
        )
        where row = 1 
        
        ),
         subsPurchased_repeat as (
          select owner_id as captain_id_repeat
          from payments.transactions_snapshot
          where lower(subtype) like '%subs%'
          and lower(owner_type)='captain'
          and yyyymmdd between '{pdd[0]}' and  '{pdd[1]}'
          and subtype = 'subscriptionPurchase'
          group by 1 
        ),
        
        byob_data as (
        
        select distinct user_id, model_type
        from (
            select 
            distinct eventprops__eventprops__userid as user_id , eventprops__eventprops__type as model_type,
            row_number() over(partition by eventprops__eventprops__userid order by yyyymmdd desc) as row
            from
              canonical.iceberg_production_data_raw_clevertap_captain_clevertap_captain_events_v2_eventname_byob_screen_immutable
            where
            yyyymmdd between '{ed[0]}' and '{ed[1]}' 
              and eventprops__eventprops__action in ('plan_confirmed')
              ) 
           where row = 1   
           
          ),
          subsCaptains as (
            select *, 
                CASE 
                WHEN captain_id_repeat IS NULL OR captain_id_repeat = '' THEN 'Previously_Not_Purchased_Subs'
            WHEN captain_id_repeat is not null THEN 'Previously_Purchased_Subs'
            ELSE 'Previously_Not_Purchased_Subs'
            END AS Subs_Purchase_or_not
            
            
            from (
               select a.*, m.captain_id_visited, 
            --   m.city,
               m.yyyymmdd, m.ingress, b.captain_id_subs as subsCaps, b.purchase_ingress,
               c.captain_id_repeat,
               m.screenName
            --   , c.counts 
              , d.user_id
              , d.model_type 
               from base a 
               left join visitedService m 
               on a.captain_id = m.captain_id_visited
               left join subs_purchased_wrt_ingress b
               on a.captain_id=b.captain_id_subs
              left join subsPurchased_repeat c 
          on a.captain_id=c.captain_id_repeat
          left join byob_data d 
          on a.captain_id = d.user_id 
            ) 
        )
        select 
        distinct * 
        from subsCaptains
    """
    df= pd.read_sql(q, metabase_connection)
    return df
```
