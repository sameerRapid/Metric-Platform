```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split


# List of variables for stratification
variables = [
    'total_login_hours', 'net_rides_taxi', 'Earnings', 'net_days', 'performance_segment', 'consistency_segment'
]

# Create buckets for each variable based on quartiles (excluding performance and consistency segments for now)
for var in variables[:-2]:  # Skip categorical variables for now
    df[f'{var}_bucket'] = pd.qcut(df[var], q=4, labels=False, duplicates='drop')

# Convert categorical variables to string codes
df['performance_segment'] = df['performance_segment'].astype(str)
df['consistency_segment'] = df['consistency_segment'].astype(str)

# Create stratification buckets for categorical variables
df['performance_segment_bucket'] = df['performance_segment'].astype('category').cat.codes
df['consistency_segment_bucket'] = df['consistency_segment'].astype('category').cat.codes

# Ensure that all stratification variables are strings and handle NaNs by replacing them with '-1'
bucket_cols = [f'{var}_bucket' for var in variables[:-2]] + ['performance_segment_bucket', 'consistency_segment_bucket']
df['stratification_bucket'] = df[bucket_cols].fillna(-1).astype(str).agg(lambda x: '-'.join(x), axis=1)

# Check if any stratification bucket has fewer than 2 members, and filter them out
counts = df['stratification_bucket'].value_counts()
valid_classes = counts[counts >= 2].index
df_filtered = df[df['stratification_bucket'].isin(valid_classes)]

# Now we will create two separate splits: 10% for test and 10% for control from the entire dataset
# Split the filtered dataset into 10% for test and 10% for control
# df_test, df_remaining = train_test_split(
#     df_filtered,
#     test_size=0.9,  # Keep 90% for remaining (test + control)
#     random_state=123,
#     stratify=df_filtered['stratification_bucket']
# )

# df_control, _ = train_test_split(
#     df_remaining,
#     test_size=0.89,  # Keep only 10% for control
#     random_state=123,
#     stratify=df_remaining['stratification_bucket']
# )
df_test, df_control = train_test_split(
    df_filtered,
    test_size=0.5,  # 50% for control
    random_state=123,
    stratify=df_filtered['stratification_bucket']
)

# Display the stratified test and control groups
print("Test Group Sample:")
print(df_test)

print("\nControl Group Sample:")
print(df_control)

# Print the stratification counts to check distribution
print("\nStratification Bucket Distribution in Test Group:")
print(df_test['stratification_bucket'].value_counts())

print("\nStratification Bucket Distribution in Control Group:")
print(df_control['stratification_bucket'].value_counts())
```
