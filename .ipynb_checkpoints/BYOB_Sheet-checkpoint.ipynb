```python
def control(sd,ed):
    query = f"""
        with base as (
        SELECT
        captain_id, city, performance_segment, consistency_segment
        -- , model_pre_exp
            --   distinct *
            from
            hive.experiments_internal.captain_byob_experiment_2_base_control
            -- hive.experiments_internal.captain_byob_experiment_2_base_test
            -- hive.experiments_internal.captain_byob_experiment_3_base_test_all_v1
           where lower(city) in ('hyderabad')
          -- where lower(city) = 'kolkata'
        --  where lower(city) = 'ahmedabad'
        ),
        vistitedcaps as (
        select  
        updated_yyyymmdd as yyyymmdd,
        updated_epoch,
        json_extract_scalar(event_props, '$.city') AS city,
        event_props_user_id as captain_id,
        json_extract_scalar(event_props, '$.screen_name') as screenName,
        json_extract_scalar(event_props, '$.source') as ingress
        from clevertap.captain_subscription_screen_visited_immutable
        where json_extract_scalar(event_props, '$.Captain_Mode') not in ('5fbe8a8a9788ac0008c4eb98','642ae204b4b6b8ec5665ce87')
        -- and yyyymmdd >= '20250206' and yyyymmdd <= '20250324' -- Hyd
        and yyyymmdd between '{ed[0]}' and '{ed[1]}'
        -- and yyyymmdd >= '20250129' and yyyymmdd <= '20250324' -- Kol
        -- and json_extract_scalar(event_props, '$.screen_name') not in ('landing_page','confirmation_sheet')
        and json_extract_scalar(event_props, '$.screen_name') not in ('confirmation_sheet') -- Landing Page
        -- and json_extract_scalar(event_props, '$.screen_name') in ('landing_page') -- Landing Page -- Lot of duplicates
        -- and lower(json_extract_scalar(event_props, '$.city')) = 'kolkata'
        -- and lower(json_extract_scalar(event_props, '$.city')) = 'ahmedabad'
         and lower(json_extract_scalar(event_props, '$.city')) in ('hyderabad')
         and event_props_user_id in (select captain_id from base)
        ),

        captainService as(
        select distinct captain_id
                from datasets_internal.captain_subservice_segments_v1
                where lower(service_mode)= 'all'
                and lower(service) = 'link'
                -- and run_date between date_format({{start_date}},'%Y%m%d') and date_format({{end_date}},'%Y%m%d')
                -- and run_date >= '20250206' and run_date <= '20250324' -- Hyd
                -- and run_date >= '20250129' and run_date <= '20250324' -- Kol
                and run_date between '{ed[0]}' and '{ed[1]}'
        ),

        visitedService as (
        select a.yyyymmdd,a.ingress,a.city,
        a.captain_id as captain_id_visited ,a.screenName,
        row_number() over(partition by a.yyyymmdd,a.captain_id,a.city order by a.updated_epoch desc) as r
        from vistitedcaps a
        inner join captainService b
        on a.captain_id=b.captain_id
        ),

        subsPurchased as (
          select 
          -- distinct owner_id as captain_id
          yyyymmdd,owner_id as captain_id, epoch
          from payments.transactions_snapshot
          where lower(subtype) like '%subs%'
          and lower(owner_type)='captain'
          -- and yyyymmdd between date_format({{start_date}},'%Y%m%d') and date_format({{end_date}},'%Y%m%d')
          -- and yyyymmdd >= '20250206' and yyyymmdd <= '20250324' -- Hyd
          --  and yyyymmdd >= '20250129' and yyyymmdd <= '20250324' -- Kol
          and yyyymmdd between '{ed[0]}' and '{ed[1]}'
          and subtype = 'subscriptionPurchase'
        ),

        subs_purchased_wrt_ingress as (
        select distinct captain_id_subs, purchase_ingress
        from (
        select b.captain_id as captain_id_subs, a.ingress as purchase_ingress, a.screenName, b.epoch, 
        row_number() over(partition by b.captain_id order by epoch desc) as row 
        from visitedService a 
        left join subsPurchased b 
        on a.captain_id_visited=b.captain_id
        and a.yyyymmdd=b.yyyymmdd
        where screenName in ('razorpay_screen','paytm_screen','cashfree_screen')
        )
        where row = 1 

        ),

        subsPurchased_repeat as (
            -- select captain_id_repeat, counts 
            -- from (
          select owner_id as captain_id_repeat
        --   owner_id as captain_id_repeat, count(*) as counts
          from payments.transactions_snapshot
          where lower(subtype) like '%subs%'
          and lower(owner_type)='captain'
          -- and yyyymmdd between date_format({{start_date}},'%Y%m%d') and date_format({{end_date}},'%Y%m%d')
          --   and yyyymmdd >= '20250101' and yyyymmdd <= '20250125' -- Kol and others 
            -- and yyyymmdd >= '20250105' and yyyymmdd <= '20250205' -- Hyd 
          and yyyymmdd between '{sd[0]}' and '{sd[1]}'
          and subtype = 'subscriptionPurchase'
          group by 1 
        ),

        byob_data as (

        select distinct user_id, model_type
        from (
            select 
            distinct eventprops__eventprops__userid as user_id , eventprops__eventprops__type as model_type,
            row_number() over(partition by eventprops__eventprops__userid order by yyyymmdd desc) as row
            -- distinct
            --   yyyymmdd,
            --   epoch,
            --   eventprops__eventprops__action as action,
            --   eventprops__eventprops__city as city,
            --   eventprops__eventprops__userid as userId,
            --   eventprops__eventprops__type as type
            from
              canonical.iceberg_production_data_raw_clevertap_captain_clevertap_captain_events_v2_eventname_byob_screen_immutable
            where
            -- yyyymmdd >= '20250206' and yyyymmdd <= '20250324' -- Hyd 
            yyyymmdd between '{ed[0]}' and '{ed[1]}'
            -- yyyymmdd >= '20250129' and yyyymmdd <= '20250324' -- Kol
              and eventprops__eventprops__action in ('plan_confirmed')
              ) 
          where row = 1   
          
          ),


        subsCaptains as (
        select *, 
            CASE 
            WHEN captain_id_repeat IS NULL OR captain_id_repeat = '' THEN 'Previously_Not_Purchased_Subs'
            WHEN captain_id_repeat is not null THEN 'Previously_Purchased_Subs'
            ELSE 'Previously_Not_Purchased_Subs'
        END AS Subs_Purchase_or_not


        from (
          select a.*, m.captain_id_visited, 
        --   m.city,
          m.yyyymmdd, m.ingress, b.captain_id_subs as subsCaps, b.purchase_ingress,
          c.captain_id_repeat,
          m.screenName
        --   , c.counts 
          , d.user_id
          , d.model_type 
          from base a 
          left join visitedService m 
          on a.captain_id = m.captain_id_visited
          left join subs_purchased_wrt_ingress b
          -- on a.captain_id = b.captain_id
          on a.captain_id=b.captain_id_subs
          -- and m.yyyymmdd=b.yyyymmdd
          left join subsPurchased_repeat c 
          on a.captain_id=c.captain_id_repeat
          left join byob_data d 
          on a.captain_id = d.user_id 
        --   where a.screenName in ('razorpay_screen','paytm_screen','cashfree_screen')
        --   and a.r=1
            ) 
          --  where captain_id = '5d7fa92ba24c96105e52706b'
          
        )
        -- ,

        select 
        distinct * 
        -- distinct Subs_Purchase_or_not, captain_id
        -- purchase_ingress, count(distinct subsCaps) as subs_caps
        from subsCaptains
      """
    return pd.read_sql(query, metabase_connection)
```

```python
test_pre_post  = pd.read_sql('''
with byob_data as (
      select distinct user_id, model_type, yyyymmdd,city
         from (
         select 
           distinct eventprops__eventprops__userid as user_id ,
           eventprops__eventprops__type as model_type,
           yyyymmdd,
           eventprops__eventprops__city as city,
           row_number() over(partition by eventprops__eventprops__userid order by yyyymmdd) as row
       from
           canonical.iceberg_production_data_raw_clevertap_captain_clevertap_captain_events_v2_eventname_byob_screen_immutable
        where  yyyymmdd <= '20250317' -- Hyd 
        and eventprops__eventprops__action in ('plan_confirmed')
      ) 
       where row = 1   
  ),
prePostTbl as (
  select b.user_id,
    b.city,
    case
       when a.yyyymmdd between b.yyyymmdd and  date_format(date_add('day',13,date_parse(b.yyyymmdd,'%Y%m%d')),'%Y%m%d') then 'Post'
       when a.yyyymmdd between  date_format(date_add('day',-14,date_parse(b.yyyymmdd,'%Y%m%d')),'%Y%m%d') and date_format(date_add('day',-1,date_parse(b.yyyymmdd,'%Y%m%d')),'%Y%m%d') then 'Pre'
    end as pre_post_flag,
    sum(cast(a.net_captain>0 as double)) as totalNetDays,
    sum(cast(a.app_open>0 as double)) as num_aoDays,
    sum(cast(a.online>0 as double)) as num_onlineDays,
    sum(cast(a.gross_captain>0 as double)) as num_grossDays,
    sum(cast(a.accepted_captain>0 as double)) as num_acceptedDays,
    sum(a.net_rides_taxi)as avg_daily_taxi_rides,
    sum(a.total_lh) as avg_lh,
    sum(a.final_captain_earnings) as avg_totalEarnings
  from byob_data b
  inner join reports_internal.consideration_temp_captain_logs_base a
  on a.captain_id=b.user_id
  and a.yyyymmdd between date_format(date_add('day',-14,date_parse(b.yyyymmdd,'%Y%m%d')),'%Y%m%d') and date_format(date_add('day',13,date_parse(b.yyyymmdd,'%Y%m%d')),'%Y%m%d')
  group by 1,2,3
)
select *
from prePostTbl
''', metabase_connection)
```
